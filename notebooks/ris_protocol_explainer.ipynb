**Interactive Simulation:** Adjust the phase shift of planet $e$ below to see how it disrupts or enhances the Great Handshake.


```python
def plot_interference(phase_shift_e=0.0, threshold=0.85):
    duration = encoder.get_great_handshake() * 2  # Plotting two handshake intervals
    t = np.linspace(0, duration, 2000)
    
    phases = {name: (freq * t) % (2 * np.pi) for name, freq in planets.items()}
    phases["TRAPPIST-1e"] = (phases["TRAPPIST-1e"] + phase_shift_e) % (2 * np.pi)
    
    # Calculate Constructive Interference
    interference = np.zeros_like(t)
    for phase_array in phases.values():
        interference += np.cos(phase_array)
        
    # Normalize 0 to 1
    interference_norm = (interference - interference.min()) / (interference.max() - interference.min())
    
    it_events = t[interference_norm > threshold]
    
    plt.figure(figsize=(12, 6))
    plt.plot(t, interference_norm, label='Resonant Interference', color='#00ffcc', alpha=0.8)
    plt.axhline(y=threshold, color='red', linestyle='--', label=f'Schrödinger Threshold ({threshold})')
    plt.scatter(it_events, [threshold]*len(it_events), color='yellow', s=15, label='"It" Events (Ignition)', zorder=5)
    
    plt.style.use('dark_background')
    plt.title("TRAPPIST-1 Constructive Interference (Mayan Gearbox)")
    plt.xlabel("Orbital Time")
    plt.ylabel("Normalized Interference Amplitude")
    plt.legend(loc='upper right')
    plt.grid(True, alpha=0.15)
    plt.show()
    
    print(f"Total 'It' Events Triggered in this window: {len(it_events)}")

# Interactive Widget
slider = widgets.FloatSlider(value=0.0, min=0.0, max=np.pi, step=0.1, description='Phase Shift (e):', layout=widgets.Layout(width='50%'))
thresh_slider = widgets.FloatSlider(value=0.85, min=0.6, max=0.99, step=0.01, description='Threshold:', layout=widgets.Layout(width='50%'))
widgets.interact(plot_interference, phase_shift_e=slider, threshold=thresh_slider);
```
